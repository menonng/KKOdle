## 📖**조건문**
- 주어진 조건이 참인지 거짓인지에 따라 다른 명령을 수행하는 구조<br>
- **⭐들여쓰기⭐**가 매우 중요하며, 이로 인한 오류에 주의해야 함<br>


### 1️⃣**기본 조건문 #if 구조**
- **if문**: 조건문을 선언하기 위한 키워드 **('만약~라면')**
- 조건이 **참**일 때, **들여쓰기 된 명령문 수행**
- 조건이 **거짓**일 때, **아무것도 수행하지 않음**<br>
<br>
**🎯조건식의 결과는 항상 참(True) 또는 거짓(False)이어야 함!**

```python
#실행시켜서 결과 확인
a = 10
if a == 10:
    print(a)
```

```python
#실행시켜서 결과 확인
a = 20
if a == 10:
    print(a)
    print('안녕')
print('끝')
```

### 2️⃣**양자 택일 #if-else 구조**

### 3️⃣**다중 택일 #if-elif-else 구조**
- **if-elif-else문**: 조건이 **여러개**일 때 사용 **('만약\~라면, 그렇지 않고 만약\~라면, 그렇지 않으면~')**<br>
- if와 else 사이에 원하는 만큼의 **elif문**을 추가할 수 있음
- **else문**은 가장 마지막 또는 생략 가능 **(❌중간에 넣을 수 없음)**<br>
<br>
- **⚠️조건식의 참, 거짓에 따라 실행되는 명령문은 1개 이거나 아예 없을 수 있음!**

```python
#실행시켜서 결과 확인
a = 10
if a < 10:
    print(f"{a} < 10")
elif a > 10:
    print(f"{a} > 10")
else:
    print(f"{a} == 10")
```

## 📖**중첩 조건문**
- 조건 안에 또 조건이 들어 있는 구조
- 어떤 조건이 참일 때, **그 안에서 다시 한 번 조건을 확인하고 세부적으로 분기 처리할 때 사용**<br>


#### **💻 입장료 프로그램[중첩조건문]**
사용자로부터 나이를 입력받아 **age 변수에 저장**하고, **8살 미만**이면 **"입장 무료"** 출력, # **~~60살 이상~~**이면 **"입장료 5000원"** 출력, **그렇지 않으면 "입장료 10000원"** 출력<br>

```python
#실행시켜서 결과 확인
age = int(input("나이 입력> "))

if age >= 8 :
    if age < 60 :
        print("입장료 10000원")  #age가 8이상이고, 60미만일 때 실행
    else :  #age가 8이상이고, 60이상일 때 실행
        print("입장료 5000원")
else :
    print("입장료 무료")  #age가 8미만일 때 실행
```

#### **💻 과자 구매 프로그램[중첩조건문]**
사용자로부터 홈런볼 가격을 입력받아 **hball 변수에 저장**하고, **money와 값을 비교**하여 **구매가 가능하고, 거스름돈이 있는 경우 "구매 가능, 거스름돈: 00원"** 출력, **구매가 가능하고, 거스름 돈이 없는 경우 "구매 가능, 거스름돈 없음"** 출력, **그렇지 않으면 "구매 불가능"** 출력<br>

```python
#실행시켜서 결과 확인
money = 5000
hball = int(input("홈런볼 가격> "))

if money >= hball :
    if money == hball :
        print(f"구매 가능, 거스름돈 없음")  #money가 hball 이상이고, money가 hball과 같을 때 실행
    else :
        print(f"구매 가능, 거스름돈: {money-hball}원") #money가 hball 이상이고, money가 hball과 다를 때 실행
else :
    print("구매 불가능")  #money가 hball 미만일 때 실행
```

#### **💻 성적 출력 프로그램[중첩조건문]**
사용자로부터 점수를 입력받아 **score 변수에 저장**하고, **80점 이상**이면 **"A등급"** 출력, **60점 이상 80점 미만**이면  **"B등급"** 출력, **40점 이상 60점 미만**이면 **"C등급"** 출력, **40점 미만**이면 **"재시험"** 출력<br>
**"시험 종료"** 출력

```python
#실행시켜서 결과 확인
score = int(input("점수 입력> "))

if score < 40 :
    print("재시험")  #score가 40 미만일 때 실행
else :
    if score < 60 :
        print("C등급")  #score가 40 이상이고, 60미만일 때 실행
    else :
        if score < 80 :
            print("B등급")  #score가 40 이상이고, 60 이상이며, 80 미만일 때 실행
        else :
            print("A등급")  #score가 40 이상이고, 60 이상이며, 80 이상일 때 실행
print("시험 종료")  
```

#### **💻 요일 판별 프로그램[중첩조건문]**
사용자로부터 요일(월, 화, 수, 목, 금, 토, 일 중 1개)을 입력 받아 **day 변수에 저장**하고, **"토" 또는 "일"** 이면 **"주말"** 출력, **"월", "화", "수", "목", "금" 중 1개** 이면 **"평일"** 출력, **그렇지 않으면 "잘못된 입력"** 출력

```python
#실행시켜서 결과 확인
day = input('요일 입력> ')

if day != '월' and day != '화' and day != '수' and day != '목' and day != '금' and day != '토' and day != '일' :
    print('잘못된 입력')  #day가 월, 화, 수, 목, 금, 토, 일 중 1개가 아닐 때 실행
else :
    if day == '토' or day =='일' :
        print('주말')  #day가 월, 화, 수, 목, 금, 토, 일 중 1개 이고, 토 또는 일일 때 실행
    else :
        print('평일')  #day가 월, 화, 수, 목, 금, 토, 일 중 1개 이고, 토 또는 일이 아닐 때 실행
```
