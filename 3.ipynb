### 📝**input().split()함수로 여러개의 값 입력받기**
- 문자열.split(구분자)<br>
  \- **구분자**를 기준으로 문자열을 나눔<br>
  \- 구분자를 생략하면 기본으로 **공백(스페이스, 탭, 줄바꿈 포함)** 을 기준으로 나눔

````python
#실행시켜서 결과 확인
#공백을 기준으로 분리하여 a, b 변수에 각각 저장
a, b = input("과일 두 개 입력> ").split()
print(a)
print(b)
````

````python
#실행시켜서 결과 확인
#,(쉼표)를 기준으로 분리하여 a, b 변수에 각각 저장
a, b = input("정수 두 개 입력> ").split(',')
print(a)
print(b)
````

#### 🤓**여기서 잠깐! 📌map()함수**
- **map(함수, 반복 가능한 값)**<br>
  \- **함수**: 적용할 함수(예: int, float 등)<br>
  \- **반복 가능한 값**: 문자열, 리스트, 튜플 등
  image.png
<br>

**⚠️int(input().split()) → 오류 발생!**<br>
　int()함수는 **문자열 하나만** 정수로 변경 가능


```
int("10")          ✅ OK
int(["10", "20"])  ❌ TypeError!
```



````python
#실행시켜서 결과 확인
a, b = map(int, input().split())
print(f"{a} + {b} = {a+b}")
print(type(a), type(b))
````

````python
#실행시켜서 결과 확인
a, b = map(float, input().split())
print(f"{a} - {b} = {a-b}")
print(type(a), type(b))
````
