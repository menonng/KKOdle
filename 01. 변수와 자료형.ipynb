## 📖**변수**
- 어떤 값이나 명령을 저장하기 위한 **기억공간(저장 공간)**
- 프로그램이 실행되는 동안 **계속해서 값을 변경할 수 있음**
- **한 번에 하나의 값만** 가질 수 있음


```python
#어떤 값이나 명령을 저장하기 위한 기억공간(저장 공간)
#실행시켜서 결과 확인
a = 10
print(a)
```

```python
#프로그램이 실행되는 동안 계속해서 값을 변경할 수 있음
#실행시켜서 결과 확인
a = 10
print(a)
a = 20
print(a)
```

```python
#한 번에 하나의 값만 가질 수 있음
#실행시켜서 결과 확인
a = 10
a = 20
a = 30
print(a)
```

## 📖**변수명 작성 규칙**
1️⃣ 영문자와 숫자, 언더바(_)를 혼합하여 사용할 수 있으며, 한글로도 변수명을 만들 수 있음<br>
　**⚠️변수명에 한글을 사용해도 오류는 발생하지 않으나, 관례상 잘 사용하지 않음<br>**
2️⃣ 첫 글자를 숫자로 시작할 수 없으며, 중간에 공백이나 특수문자(!,~,^,$ 등)가 들어가면 안됨<br>
3️⃣ 영어의 대/소문자를 구별함(A와 a는 다른 변수)<br>
4️⃣ 프로그램 내의 특별한 **키워드(역할이 있는 명령어)** 를 사용하면 안됨<br>
5️⃣ 변수의 저장된 값을 유추할 수 있는 이름으로 설정하는 것이 좋음 **(ex. 이름을 저장하는 변수 이름 name)**

## 📖**자료형**
- **변수에 저장되는 데이터의 형식**
- **데이터형**이라고도 함

```python
#정수형
#실행시켜서 결과 확인
x = 10
print(x)
print(type(x))
```

### 🤓**여기서 잠깐! 📌type() 함수**
- 괄호 안에 들어가는 **변수나 데이터의 자료형을 알려주는 역할**

```python
#실수형
#실행시켜서 결과 확인
x = 3.14
y = 6/2
print(x, y) #정수를 정수로 나눈 결과는 실수가 됨!
print(type(x), type(y))
```

```python
#문자열형①
#실행시켜서 결과 확인
a = 'x'
b = 'Hello World'
c = "정보"

print(a)
print(b)
print(c)
print(type(a), type(b), type(c))
```

```python
#문자열형②
#실행시켜서 결과 확인
a = 30
print(a)
print(type(a))

b = '30'
print(b)
print(type(b))
```

### 🤓**여기서 잠깐! 📌30과 '30'의 차이**
- 따옴표의 유무
- 30 → **정수형 숫자**로 인식
- '30' → **문자열형 문자**로 인식(⭐따옴표로 감싸져있기 때문!)

```python
#논리형(부울형)①
#실행시켜서 결과 확인
a = True
b = False
c = 'True'

print(a)
print(b)
print(c)
print(type(a), type(b), type(c))
```

```python
#논리형(부울형)②
#실행시켜서 결과 확인
a = 10 > 20
b = 10 < 20

print(a)
print(b)
print(type(a), type(b))
```

### 📝**자료형 변환하기**
- **변환하고자 하는 자료형(변수(또는 데이터))**
- 예: int('10') #원래 문자열형이었던 '10'을 정수형(int) 숫자로 변환

```python
#실행시켜서 결과 확인
x = 10
y = '20'
print(x, y)
print(type(x), type(y))

print(type(str(x)), type(int(y)))
print(type(x), type(y))  #기존에 저장된 x, y값은 바뀌지 않음!

x = str(x)  #x의 값을 문자열형(str)으로 바꿔 다시 x에 저장
y = int(y)  #y의 값을 정수형(int)으로 바꿔 다시 y에 저장

print(x, y)
print(type(x), type(y))
```
