### 📝**문자열의 인덱싱과 슬라이싱**

#### **1️⃣인덱싱(Indexing): 무엇인가를 '가리킨다'의 의미**
image.png

```python
#실행시키기
#문자열 생성
s = "안녕하세요!"
```

#### 🤓**여기서 잠깐! 📌인덱스 번호는 중괄호([ ])로 감싸서 표현!**

```python
#3번 인덱스 위치에 있는 문자 출력
#실행시켜서 결과 확인
print(s[3])
```

```python
#-5번 인덱스 위치에 있는 문자 출력
#실행시켜서 결과 확인
print(s[-5])
```

```python
#0, 5번 인덱스 위치에 있는 문자 한 줄로 출력
#실행시켜서 결과 확인
print(s[0], s[5])
```

#### **2️⃣슬라이싱(Slicing): 무엇인가를 '잘라낸다'의 의미**
image.png

```python
#문자열 생성
#실행시키기
s = "안녕하세요!"
```

#### 🤓**여기서 잠깐! 📌슬라이싱 기본 문법**
image.png

```python
#문자열[시작 인덱스:끝 인덱스] → 시작 인덱스(포함)부터 끝 인덱스(미포함) 까지 출력
#실행시켜서 결과 확인
print(s[1:4])
print(s[3:5])
print(s[-4:-1])
```

```python
#문자열[시작 인덱스:] → 시작 인덱스(포함)부터 끝까지 출력
#실행시켜서 결과 확인
print(s[0:])
print(s[2:])
```

```python
#문자열[:끝 인덱스] → 처음부터 끝 인덱스(미포함)까지 출력
#실행시켜서 결과 확인
print(s[:4])
print(s[:3])
```

```python
#문자열[:] → 전체 복사
#실행시켜서 결과 확인
print(s[:])
print(s)
```

```python
#문자열[시작 인덱스:끝 인덱스: 간격] → 시작 인덱스(포함)부터 끝 인덱스(미포함)까지 간격에 따라 출력
#실행시켜서 결과 확인
print(s[1:5:2])
print(s[0:6:2])
```

#### 🚫**인덱스 범위를 벗어나도 에러는 안남**

```python
#범위를 벗어나도 가능한 범위까지만 출력됨
#실행시켜서 결과 확인
print(s[0:100])
```

#### **1️⃣len(문자열): 문자열의 길이 반환**

#### **2️⃣문자열.upper(): 문자열을 모두 대문자로 변환**

```python
#실행시켜서 결과 확인
print("abc".upper())
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s)
print(s.upper())
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s)
print(s.upper())

s = s.upper()
print(s)
print(s.upper())
```

#### **3️⃣문자열.lower(): 문자열을 모두 소문자로 변환**

```python
#실행시켜서 결과 확인
print("ABC".lower())
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s)
print(s.lower())
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s)
print(s.lower())

s = s.lower()
print(s)
print(s.lower())
```

#### **4️⃣문자열.find(특정 문자열): 문자열에서 특정 문자열의 위치(인덱스) 반환**
- 문자열에 특정 문자열이 여러개 존재하는 경우, **가장 앞에 있는 문자열의 위치(인덱스) 반환**
- ⚠️존재하지 않으면 **-1** 반환
- **안전하게 실패할 수 있음** → 조건문에서 많이용

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.find('o'))
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.find('!'))
```

#### **5️⃣문자열.index(특정 문자열): 문자열에서 특정 문자열의 위치(인덱스) 반환**
- 문자열에 특정 문자열이 여러개 존재하는 경우, **가장 앞에 있는 문자열의 위치(인덱스) 반환**
- ⚠️존재하지 않으면 **ValueError** 오류 발생
- **무조건 있어야 하는 값**을 찾을 때 유용 → 없으면 에러 

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.index('o'))
```

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.index('!'))
```

#### **6️⃣문자열.count(특정 문자열): 문자열에서 특정 문자열이 등장하는 횟수 반환**

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.count('o'))
```

#### **7️⃣문자열.replace(특정 문자열, 다른 문자열): 문자열에서 특정 문자열을 다른 문자열로 대체**

```python
#실행시켜서 결과 확인
s = "Hello World"
print(s.replace('o', 'a'))
```
