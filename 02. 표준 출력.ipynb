### 🤓**여기서 잠깐! 📌함수의 의미**
- **수학**에서 함수: **하나의 입력에 대해 딱 하나의 출력이 정해지는 규칙**<br>
　ex) f(x) = x + 2 라는 함수는, x에 3을 넣으면 5가 나오는 식
- **컴퓨터 언어**에서 함수: **필요한 작업을 묶어 놓은 명령어 모음**<br>
　ex) "숫자 두 개를 더해주는 코드 묶음"을 만들어서 필요할 때마다 불러서 사용

### 3️⃣**문자열 포맷팅(%)을 이용한 출력**



```python
#정수형 숫자 출력
#실행시켜서 결과 확인
a = 2
b = 3
print("%d + %d = %d" %(a, b, a+b))
print("%02d + %02d = %02d" %(a, b, a+b))
```

```python
#실수형 숫자 출력
#실행시켜서 결과 확인
a = 3.0
b = 0.14
print("%f + %f = %f" %(a, b, a+b))
print("%.3f + %.1f = %.2f" %(a, b, a+b))
```

```python
#문자열 출력
#실행시켜서 결과 확인
a = "안녕"
b = "하세요"
print("%s + %s = %s" %(a, b, a+b))
```

```python
#단일 문자 출력
#실행시켜서 결과 확인
a = "H"
b = "i"
print("%c%c" %(a, b))
```

### 4️⃣**format()을 이용한 출력**


⚠️format()을 이용하면 변수의 자료형에 상관없이 문자열을 포맷팅 할 수 있다는 장점이 있으나, 코드가 길어지는 단점이 있음

```python
#실행시켜서 결과 확인
a = 10
b = 20
print("{} + {} = {}".format(a, b, a+b))
```

```python
#실행시켜서 결과 확인
a = "안녕"
b = "하세요"
print("{} + {} = {}".format(a, b, a+b))
```

### 6️⃣**키워드 sep을 이용한 출력**
- 키워드 sep을 이용하면 각 항목(변수 또는 데이터) 사이에 **특정 문자열**이나 **기호**를 넣을 수 있음
- 키워드 sep을 따로 설정하지 않으면 각 항목 사이에는 **공백**이 삽입됨

```python
#실행시켜서 결과 확인
year = 2025
month = 5
day = 16
print(year, month, day)
print(year, month, day, sep='/')
print("끝")
```

### 7️⃣**키워드 end를 이용한 출력**
- 키워드 end는 **출력되는 내용의 마지막에 들어갈 문자열을 정의**하는데 사용됨

```python
#실행시켜서 결과 확인
year = 2025
month = 5
day = 16
print(year, month, day)
print(year, month, day, end='/')
print("끝")
print("끝2")
```

## 📖**이스케이프 문자 #\\(=￦): 백 슬래시**
- \ 기호와 함께 조합해서 사용하는 특수한 문자

```python
#실행시켜서 결과 확인
print('"안녕하세요"')
print("\"안녕하세요\"")
```

```python
#실행시켜서 결과 확인
print("'안녕하세요'")
print("\'안녕하세요\'")
```

```python
#실행시켜서 결과 확인
print("정보 시간\n즐거운 프로그래밍")
print("python")
```

```python
#실행시켜서 결과 확인
print("정보 시간\t즐거운 프로그래밍")
```

```python
#실행시켜서 결과 확인
print("1학년\t정보\t프로그래밍")
```
